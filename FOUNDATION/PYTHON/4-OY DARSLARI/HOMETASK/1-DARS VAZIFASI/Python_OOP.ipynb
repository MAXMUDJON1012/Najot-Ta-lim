```python
# 1- Task

class Phone:
    def __init__(self, model, memory, battery):
        self.model = model
        self.memory = memory
        self.battery = battery

    def info(self):
        print(f"Telefon modeli: {self.model}, Xotirasi: {self.memory}, Batereya sig'imi {self.battery} mAh")
  
    def zaryad_etarliymi(self):
        using_time = int(input("Enter using time: "))
        if (self.battery / 200 >= using_time):
            print(f"Siz telefondann foydalanishiz mumkun {self.battery / 200} shuncha voqt")
        else:
            print(f"Sizga shuncha {(using_time * 200) - self.battery} zaryad kerak, buncha {using_time} vaqt ishlatishiz uchun.")

phone1 = Phone("S26 Ultra", "1 T", 5000) 
phone1.info()
phone1.zaryad_etarliymi()
```

```python
# 3-TASK
class Server:

    def __init__(self, nomi, ram, disk):
        self.nomi = nomi
        self.ram = ram
        self.disk = disk

    def texnik_malumot(self):
        print(f"Server: {self.nomi} \nRAM: {self.ram} GB \nDisk:{self.disk} TB")

    def saralash(self, fayl_hajmi):
        disk_gb = self.disk * 1024
        if fayl_hajmi <= disk_gb:
            print(f"{fayl_hajmi} GB fayl {self.disk} TB diskka sig'adi!")
        else:
            print(f"{fayl_hajmi} GB fayl {self.disk} TB diskka sig'maydi!")


s = Server("MainServer",64,2)
s.texnik_malumot()
s.saralash(1500)
```

```python
# 4-TASK


class Vebsayt:
    
    def __init__(self, domen, til, oylik_tashrif):
        self.domen = domen
        self.til = til
        self.oylik_tashrif = oylik_tashrif

    def info(self):
        print(f"Domen: {self.domen} \nTexnologiya: {self.til} \nTashriflar soni: {self.oylik_tashrif}")

    def mashhurlik(self):
        if self.oylik_tashrif < 10000:
            daraja = "Boshlang'ich"
        elif 10000 <= self.oylik_tashrif <= 100000:
            daraja = "O'rta"
        else:
            daraja = "Mashhur"
        
        print(f"{self.domen} — {daraja} sayt")


v = Vebsayt("uzum.uz", "React", 250000)
v.info()
v.mashhurlik()
```

```python
# 5-TASK
class Bugtracker:
    def __init__(self, loyiha_nomi, xatolar, dasturchi):
        self.loyiha_nomi = loyiha_nomi
        self.xatolar = xatolar
        self.dasturchi = dasturchi

    def xato_qoshish(self, xato):
        self.xatolar.append(xato)

    def hisobot(self):
        print(f"Loyiha: {self.loyiha_nomi} \nDasturchi: {self.dasturchi}")
        print("Xatolar:")
        
        nomer = 1
        for xato in self.xatolar:
            print(f"{nomer}. {xato}")
            nomer += 1


b = Bugtracker("OnlineShop", [], "Vali")
b.xato_qoshish("Login sahifasi ishlamaydi")
b.xato_qoshish("To'lov xatosi")
b.hisobot()
```

```python
class Hayvon:
    def __init__(self, ism, yosh):
        self.ism = ism
        self.yosh = yosh

    def info(self):
        print(f"Ismi: {self.ism}, Yoshi: {self.yosh}")


class It(Hayvon):
    def vovulla(self):
        print("Vov-vov!")


class Mushuk(Hayvon):
    def miyovla(self):
        print("Miyov-miyov!")


it = It("Baron", 3)
mushuk = Mushuk("Mioshvoy", 2)

it.info()
it.vovulla()

mushuk.info()
mushuk.miyovla()
```

```python
class Transport:
    def __init__(self, nomi, tezlik):
        self.nomi = nomi
        self.tezlik = tezlik

    def harakat(self):
        print(f"{self.nomi} {self.tezlik} km/h tezlikda harakatlanmoqda.")


class Avtomobil(Transport):
    def __init__(self, nomi, tezlik, rusumi):
        super().__init__(nomi, tezlik)
        self.rusumi = rusumi

    def harakat(self):
        super().harakat()
        print(f"Rusumi: {self.rusumi}")


class Velosiped(Transport):
    def __init__(self, nomi, tezlik, turi):
        super().__init__(nomi, tezlik)
        self.turi = turi

    def harakat(self):
        super().harakat()
        print(f"Turi: {self.turi}")


avto = Avtomobil("KIA K5", 180, "GT")
avto.harakat()

velo = Velosiped("Jeep p30", 25, "Tog'")
velo.harakat()
```

```python
import math

class Shakl:
    def __init__(self, rang):
        self.rang = rang

    def yuza(self):
        return 0


class Kvadrat(Shakl):
    def __init__(self, rang, tomon):
        super().__init__(rang)
        self.tomon = tomon

    def yuza(self):
        return self.tomon ** 2


class Aylana(Shakl):
    def __init__(self, rang, radius):
        super().__init__(rang)
        self.radius = radius

    def yuza(self):
        return math.pi * (self.radius ** 2)


class Uchburchak(Shakl):
    def __init__(self, rang, asos, balandlik):
        super().__init__(rang)
        self.asos = asos
        self.balandlik = balandlik

    def yuza(self):
        return 0.5 * self.asos * self.balandlik


kv = Kvadrat("Qizil", 4)
print(kv.rang, kv.yuza())

ay = Aylana("Ko'k", 3)
print(ay.rang, round(ay.yuza(), 2))

uch = Uchburchak("Yashil", 6, 4)
print(uch.rang, uch.yuza())
```

```python
class Xodim:
    def __init__(self, ism, maosh):
        self.ism = ism
        self.maosh = maosh

    def info(self):
        print(f"Ism: {self.ism}, Maosh: {self.maosh}")

    def bonus(self):
        return self.maosh * 0.10


class Dasturchi(Xodim):
    def __init__(self, ism, maosh, til):
        super().__init__(ism, maosh)
        self.til = til

    def bonus(self):
        return self.maosh * 0.20

    def info(self):
        super().info()
        print(f"Til: {self.til}")


class Dizayner(Xodim):
    def __init__(self, ism, maosh, dastur):
        super().__init__(ism, maosh)
        self.dastur = dastur

    def bonus(self):
        return self.maosh * 0.15

    def info(self):
        super().info()
        print(f"Dastur: {self.dastur}")


class Menejer(Xodim):
    def __init__(self, ism, maosh, bolim):
        super().__init__(ism, maosh)
        self.bolim = bolim

    def bonus(self):
        return self.maosh + 500

    def info(self):
        super().info()
        print(f"Bo'lim: {self.bolim}")


xodimlar = [
    Dasturchi("Ali", 2000, "Python"),
    Dizayner("Vali", 1500, "Figma"),
    Menejer("Sami", 3000, "Sotuv")
]

for xodim in xodimlar:
    xodim.info()
    print(xodim.bonus())
```

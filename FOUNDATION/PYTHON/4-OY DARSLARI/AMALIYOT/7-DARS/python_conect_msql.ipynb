```python
import mysql.connector

mydb = mysql.connector.connect(
    host="localhost",
    user="root",
    password="2005",
    database="school"
)

cursor = mydb.cursor()

fullname = input("Enter fullname: ")
age = int(input("Enter age: "))
phone = input("Enter phone number: ")
score = float(input("Enter score: "))ali
sql = """
INSERT INTO students (fullname, age, phone, score)
VALUES (%s, %s, %s, %s)
"""

values = (fullname, age, phone, score)

cursor.execute(sql, values)
mydb.commit()

print("✅ Student successfully added!")
```

```python
class SSD:
    def __init__(self,nomi,hajmi):
        self.nomi=nomi
        self.hajmi=hajmi

    def __add__(self, other):
        assert isinstance   (other,SSD), ValueError("Faqat SSD larni qo'shsa bo'ladi")
        self.hajmi +=other.hajmi
        self.nomi=other.nomi
        return self

    def __eq__(self, value):
        if isinstance(value,SSD):
            return self.hajmi == value.hajmi
        return self.hajmi == value
    
    def __len__(self):
        return self.hajmi

    def __str__(self):
        return f"{self.nomi} {self.hajmi} GB"

ssd1=SSD("Sotvoldihoji's",1024)
ssd2=SSD("Muhammadaziz's",512)
    #  print(ssd1+ssd2)

print(ssd2 == 1024)
print(ssd1 == 1024)
print(ssd1 == ssd2)

print(len(ssd1))
```

```python
class Laptop:
    def __init__(self,brand, ram, ssd, price):
        self.brand=brand
        self.ram=ram
        self.ssd=ssd
        self.price=price

    def __str__(self):
        return f"Laptop(brand = {self.brand}, ram = {self.ram}GB, ssd = {self.ssd}GB, price = ${self.price})"

    def __getattr__(self, name):
        assert name in ["brand","ram","ssd","price"], AttributeError(f"Bunday atribut yo'q ->{name}")
        return self.name

    def __setattr__(self, name, value):
        if not name in ["brand","ram","ssd","price"]:
            raise AttributeError(f"Bunday atribut yo'q ->{name}")
        super().__setattr__(name, value)

l1=Laptop("MacBook",16,512,950)
print(l1)
print(l1.brand)
l1.ssd=256
# l1.yadro=10
# print(l1.yadro)
```

```python
class Car:
    def __init__(self,name,motor):
        self.name=name
        self.motor=motor


class K5_GT(Car):
    def __init__(self,name,motor,collor):
        super().__init__(name,motor)
        self.collor=collor
    def mator(self):
        print(f"{self.name} mashina {self.motor} Turbo mator rangi  {self.collor}")

car1 = K5_GT("K5_GT", 3.2, "red")
car1.mator()
```

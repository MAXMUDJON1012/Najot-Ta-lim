```python

class Kitob:
    def __init__(self, nomi, author, sahifalar_soni=0):
        self.nomi = nomi
        self.author = author
        self.sahifalar_soni = sahifalar_soni

    def oqildi(self):
        return f"{self.nomi} bu kitob o'qib bo'lindi"
    
    def hajm(self):
        if self.sahifalar_soni > 250:
            return f"{self.nomi} bu kitobni o'qisa bo'laadi"
        else:
            return "o'qish qiyin"
    def malumot(self):
        return f"Kitob nomi:{self.nomi}\nMualif:{self.author}\nSaxifalar soni:{self.sahifalar_soni}"
    

kitob1 = Kitob("Ikki eshik orasi", "O'tkir Hoshimov", 350)
kitob2 = Kitob("Oq kema", "Chingiz Aytmatov")
kitob3 = Kitob("Ufq", "Said Ahmad")
print(kitob1.oqildi())
print(kitob2.oqildi())
print(kitob3.oqildi())
# print(kitob1.hajm())
# print(kitob1.malumot())
books=[kitob1,kitob2,kitob3]



for b in books:
    print(b.malumot)
```

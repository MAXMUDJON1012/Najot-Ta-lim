```python
class BankHisobi:
    def __init__(self, egasi, balans=0):
        self.__egasi = egasi
        self.__balans = balans

    def pul_qosh(self, miqdor):
        if miqdor > 0:
            self.__balans += miqdor
            print(f"{miqdor} so'm qo'shildi.")

    def pul_yech(self, miqdor):
        if miqdor <= self.__balans:
            self.__balans -= miqdor
            print(f"{miqdor} so'm yechildi.")
        else:
            print("Balansda mablag' yetarli emas!")

    def balansni_kor(self):
        print(f"Egasi: {self.__egasi} | Balans: {self.__balans} so'm")


hisob = BankHisobi("Ali", 100000)
hisob.pul_qosh(50000)
hisob.pul_yech(30000)
hisob.pul_yech(200000)
hisob.balansni_kor()
```

```python
class Talaba:
    def __init__(self, ism, yosh, baho=0):
        self.__ism = ism
        self.__yosh = yosh
        self.__baho = 0
        self.bahoni_ornat(baho)

    def bahoni_ornat(self, baho):
        if 0 <= baho <= 100:
            self.__baho = baho
        else:
            print("Noto'g'ri baho! Baho 0 dan 100 gacha bo'lishi kerak.")

    def bahoni_kor(self):
        return self.__baho

    def malumotni_kor(self):
        print(f"Ism: {self.__ism}, Yosh: {self.__yosh}, Baho: {self.__baho}")


talaba = Talaba("Vali", 20, 85)
talaba.malumotni_kor()
talaba.bahoni_ornat(105)
talaba.bahoni_ornat(95)
print("Baho:", talaba.bahoni_kor())
```

```python
from abc import ABC, abstractmethod


class Shakl(ABC):
    @abstractmethod
    def yuzani_hisobla(self):
        pass

    @abstractmethod
    def perimetrni_hisobla(self):
        pass


class Tortburchak(Shakl):
    def __init__(self, uzunlik, en):
        self.uzunlik = uzunlik
        self.en = en

    def yuzani_hisobla(self):
        return self.uzunlik * self.en

    def perimetrni_hisobla(self):
        return 2 * (self.uzunlik + self.en)


uzunlik = float(input("Uzunlikni kiriting: "))
en = float(input("Enini kiriting: "))

tb = Tortburchak(uzunlik, en)
print(f"Yuzi: {tb.yuzani_hisobla()}")
print(f"Perimetri: {tb.perimetrni_hisobla()}")
```

```python
from abc import ABC, abstractmethod


class Xodim(ABC):
    @abstractmethod
    def maoshni_hisobla(self):
        pass


class Dasturchi(Xodim):
    def __init__(self, ism, oylik):
        self.__ism = ism
        self.__oylik = oylik

    def maoshni_hisobla(self):
        yillik = self.__oylik * 12
        print(f"{self.__ism}ning yillik maoshi: {yillik}$")
        return yillik


dasturchi = Dasturchi("Sardor", 1500)
dasturchi.maoshni_hisobla()
```

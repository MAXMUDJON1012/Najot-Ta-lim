```python
class Kitob:
    def __init__(self, nomi, muallif, sahifalar):
        self.nomi = nomi
        self.muallif = muallif
        self.sahifalar = sahifalar

    def __str__(self):
        return f"{self.nomi} ({self.muallif}) — {self.sahifalar} sahifa"

    def __len__(self):
        return self.sahifalar

    def __eq__(self, other):
        if isinstance(other, Kitob):
            return self.sahifalar == other.sahifalar
        return self.sahifalar == other

    def __lt__(self, other):
        if isinstance(other, Kitob):
            return self.sahifalar < other.sahifalar
        return self.sahifalar < other

    def __gt__(self, other):
        if isinstance(other, Kitob):
            return self.sahifalar > other.sahifalar
        return self.sahifalar > other


kitoblar = [
    Kitob("Xamsa", "Navoiy", 850),
    Kitob("Sariq devni minib", "Xudoyberdi To'xtaboyev", 240),
    Kitob("O'tkan kunlar", "Abdulla Qodiriy", 350)
]

saralangan = sorted(kitoblar)

for kitob in saralangan:
    print(kitob)
```

```python
class Temperatura:
    def __init__(self, daraja, birlik):
        self.daraja = daraja
        self.birlik = birlik

    def __str__(self):
        return f"{self.daraja}°{self.birlik}"

    def __repr__(self):
        return f"Temperatura({self.daraja}, '{self.birlik}')"

    def __eq__(self, other):
        val = other.daraja if isinstance(other, Temperatura) else other
        return self.daraja == val

    def __lt__(self, other):
        val = other.daraja if isinstance(other, Temperatura) else other
        return self.daraja < val

    def __le__(self, other):
        val = other.daraja if isinstance(other, Temperatura) else other
        return self.daraja <= val

    def __gt__(self, other):
        val = other.daraja if isinstance(other, Temperatura) else other
        return self.daraja > val

    def __ge__(self, other):
        val = other.daraja if isinstance(other, Temperatura) else other
        return self.daraja >= val


t1 = Temperatura(36.6, "C")
t2 = Temperatura(38.0, "C")

print(str(t1))
print(repr(t1))
print(t1 < t2)
print(t1 >= t2)
```

```python
class Talaba:
    def __init__(self, ism, baholar):
        self.ism = ism
        self.baholar = baholar

    def __str__(self):
        fanlar_soni = len(self.baholar)
        ortacha = sum(self.baholar.values()) / fanlar_soni if fanlar_soni > 0 else 0
        return f"{self.ism} — {fanlar_soni} ta fan, o'rtacha baho: {ortacha:.1f}"

    def __len__(self):
        return len(self.baholar)

    def __setattr__(self, name, value):
        if name == "baholar" and isinstance(value, dict):
            for fan, baho in value.items():
                if not (0 <= baho <= 100):
                    raise ValueError(f"Baho 0–100 oraliqda bo'lishi kerak: {fan} -> {baho}")
        super().__setattr__(name, value)

    def __getattr__(self, name):
        return f"Bu attribute mavjud emas: {name}"


t = Talaba("Ali", {"Matematika": 90, "Fizika": 95, "Tarix": 90})

print(t)
print(len(t))
print(t.yosh)
```
